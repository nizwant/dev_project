```python
import matplotlib.pyplot as plt
import random
import numpy as np
import pandas as pd
import seaborn as sns
from collections import defaultdict

from sklearn.decomposition import PCA, TruncatedSVD, FactorAnalysis
from sklearn.manifold import Isomap
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
from sklearn.feature_selection import mutual_info_classif
from sklearn.linear_model import LassoCV, LogisticRegression
from sklearn.metrics import log_loss
from sklearn.ensemble import RandomForestClassifier

import sys

sys.path.append("./..")

from src.generate_data import generate_dataset
```

```python
def select_features(X_train, X_test, columns, key, n_components=12):
    if key == "pca":
        pca = PCA(n_components=n_components)
        X_train_transformed = pca.fit_transform(X_train)
        X_test_transformed = pca.transform(X_test)
        return X_train_transformed, X_test_transformed

    elif key == "svd":
        svd = TruncatedSVD(n_components=n_components, random_state=42)
        X_train_transformed = svd.fit_transform(X_train)
        X_test_transformed = svd.transform(X_test)
        return X_train_transformed, X_test_transformed

    elif key == "fa":
        fa = FactorAnalysis(n_components=n_components, random_state=42)
        X_train_transformed = fa.fit_transform(X_train)
        X_test_transformed = fa.transform(X_test)
        return X_train_transformed, X_test_transformed

    elif key == "isomap":
        isomap = Isomap(n_components=n_components)
        X_train_transformed = isomap.fit_transform(X_train)
        X_test_transformed = isomap.transform(X_test)
        return X_train_transformed, X_test_transformed

    if key not in columns:
        raise Exception("key not in dictionary")
    return X_train[:, columns[key]], X_test[:, columns[key]]
```

```python
def run_selection_methods(X_train, y_train, feature_types, seed, silent=True):
    num_of_informative_features = len(feature_types["informative"])

    # Store results
    selected_features = {
        "all features": list(range(20)),
        "informative only": feature_types["informative"],
    }

    # --- 1. LASSO ---
    lasso = LassoCV(cv=5, random_state=seed, max_iter=10000).fit(X_train, y_train)
    lasso_selected = np.where(lasso.coef_ != 0)[0].tolist()
    selected_features["lasso"] = lasso_selected

    # --- 2. Mutual Information ---
    mi_scores = mutual_info_classif(X_train, y_train, random_state=seed)
    mi_selected = np.argsort(mi_scores)[-num_of_informative_features:].tolist()
    selected_features["mutual information"] = mi_selected

    # --- 3. Correlation ---
    correlations = []
    for i in range(X_train.shape[1]):
        corr = np.corrcoef(X_train[:, i], y_train)[0, 1]
        correlations.append((i, abs(corr)))

    correlated = [
        i
        for i, _ in sorted(correlations, key=lambda x: -x[1])[
            :num_of_informative_features
        ]
    ]
    selected_features["correlation"] = correlated

    # --- 4. BIC Wrapper (Logistic Regression with feature subsets) ---
    def bic_score(model, X, y):
        """Calculate BIC for a fitted logistic regression model"""
        n = len(y)
        k = X.shape[1]
        prob = model.predict_proba(X)[:, 1]
        ll = -log_loss(y, prob, normalize=False)  # log-likelihood
        bic = k * np.log(n) - 2 * ll
        return bic

    def bic_forward_selection(X, y, max_features=None):
        selected = []
        remaining = list(range(X.shape[1]))
        best_bic = np.inf

        while remaining:
            scores = []
            for feat in remaining:
                candidate = selected + [feat]
                model = LogisticRegression(solver="liblinear", max_iter=10000).fit(
                    X[:, candidate], y
                )
                bic = bic_score(model, X[:, candidate], y)
                scores.append((bic, feat))

            # Find the best new feature to add
            scores.sort()
            best_new_bic, best_feat = scores[0]

            # Stop if BIC does not improve
            if best_new_bic < best_bic:
                best_bic = best_new_bic
                selected.append(best_feat)
                remaining.remove(best_feat)
                if max_features and len(selected) >= max_features:
                    break
            else:
                break

        return selected

    bic_selected = bic_forward_selection(X_train, y_train)
    selected_features["bic"] = bic_selected

    # --- 5. Random Forest Feature Importance ---
    rf = RandomForestClassifier(n_estimators=100, random_state=seed)
    rf.fit(X_train, y_train)

    importances = rf.feature_importances_
    rf_selected = np.argsort(importances)[-num_of_informative_features:].tolist()
    selected_features["random forest"] = rf_selected

    # --- Print results ---
    if not silent:
        print()
        for method, features in selected_features.items():
            print(f"{method}: selected feature indices: {features}")

    return selected_features
```

```python
svm_accuracy = defaultdict(list)
svm_precision = defaultdict(list)
svm_recall = defaultdict(list)
svm_f1 = defaultdict(list)

rf_accuracy = defaultdict(list)
rf_precision = defaultdict(list)
rf_recall = defaultdict(list)
rf_f1 = defaultdict(list)
```

```python
rng = random.Random(213)

for i in range(50):
    seed = rng.randint(1, 10000)
    X, y, feature_types = generate_dataset(
        n_samples=2000, n_features=20, random_state=seed
    )

    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=0.5, random_state=42
    )

    selected_features = run_selection_methods(X_train, y_train, feature_types, seed)
    selected_features["pca"] = None
    selected_features["svd"] = None
    selected_features["fa"] = None
    selected_features["isomap"] = None

    for key in selected_features:
        X_prim_train, X_prim_test = select_features(
            X_train, X_test, selected_features, key
        )

        # Train classifiers
        svm_clf = SVC(kernel="rbf", C=1.0, gamma="scale", random_state=42)
        svm_clf.fit(X_prim_train, y_train)

        rf_clf = RandomForestClassifier(n_estimators=100, random_state=42)
        rf_clf.fit(X_prim_train, y_train)

        # Predictions
        svm_preds = svm_clf.predict(X_prim_test)
        rf_preds = rf_clf.predict(X_prim_test)

        # SVM metrics
        svm_accuracy[key].append(accuracy_score(y_test, svm_preds))
        svm_precision[key].append(precision_score(y_test, svm_preds, average="binary"))
        svm_recall[key].append(recall_score(y_test, svm_preds, average="binary"))
        svm_f1[key].append(f1_score(y_test, svm_preds, average="binary"))

        # RF metrics
        rf_accuracy[key].append(accuracy_score(y_test, rf_preds))
        rf_precision[key].append(precision_score(y_test, rf_preds, average="binary"))
        rf_recall[key].append(recall_score(y_test, rf_preds, average="binary"))
        rf_f1[key].append(f1_score(y_test, rf_preds, average="binary"))
```

```python
df = pd.DataFrame(
    [
        {"Method": k, "Accuracy": val}
        for k, values in svm_accuracy.items()
        for val in values
    ]
)

# Plot
plt.figure(figsize=(12, 6))
sns.boxplot(data=df, x="Method", y="Accuracy")
plt.xticks(rotation=45, ha="right")
plt.title(
    "SVM Accuracy on test set comparison across feature selection methods over 50 runs"
)
plt.tight_layout()
plt.show()
```

```python
df = pd.DataFrame(
    [
        {"Method": k, "Accuracy": val}
        for k, values in rf_accuracy.items()
        for val in values
    ]
)

# Plot
plt.figure(figsize=(12, 6))
sns.boxplot(data=df, x="Method", y="Accuracy")
plt.xticks(rotation=45, ha="right")
plt.title(
    "Random Forrest Accuracy on test set comparison across feature selection methods over 50 runs"
)
plt.tight_layout()
plt.show()
```
